```python
import pandas as pd
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/employees_system_dataset.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
new_df = df.dropna()

print(new_df.to_string())
```

```python
df.dropna(inplace = True)

print(df.to_string())
```

```python
df.fillna(130, inplace = True)
```

```python
print(df.duplicated())
```

```python
df.drop_duplicates(inplace = True)
```

```python
# ============================================================
# Create salary rank within each department
# Highest salary gets Rank = 1
# ============================================================

df["salary_rank"] = (
    df
    .groupby("department")["salary"]
    .rank(
        ascending=False,
        method="dense"
    )
)
```

```python
df["salary_rank"]
```

```python
# ============================================================
# Sort by department and salary rank
# ============================================================

result = df.sort_values(
    by=["department", "salary_rank"]
)
```

```python
print(result)
```

# Find the top 3 highest paid employees from every department.

```python
# Sort salary from highest to lowest
sorted_df = df.sort_values(
    by="salary",
    ascending=False
)
```

```python
# Take top 3 employees from each department
top3 = (
    sorted_df
    .groupby("department")
    .head(3)
    .reset_index(drop=True)
)
```

```python
# Display required columns
print(
    top3[
        [
            "department",
            "full_name",
            "salary"
        ]
    ]
)
```
