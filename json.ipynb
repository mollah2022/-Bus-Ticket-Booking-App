```python
# Filter employees whose salary is greater than 70000
df[df["salary"] > 70000]
```

```python
# Filter employees from AI department
df[df["department"] == "AI"]
```

```python
# Exclude AI department
df[df["department"] != "AI"]
```

```python
# Salary greater than or equal to 70000

df[df["salary"] >= 70000]
```

```python
# Salary less than 70000
df[df["salary"] < 70000]
```

```python
# AI employees whose salary is greater than 90000

df[
    (df["department"] == "AI")
    &
    (df["salary"] > 90000)
]
```

```python
# Employees from AI or BI department

df[
    (df["department"] == "AI")
    |
    (df["department"] == "BI")
]
```

```python
# Filter employees from selected departments

df[
    df["department"].isin(
        ["AI", "BI"]
    )
]
```

```python
# Employees whose salary is between 65000 and 90000
df[
    df["salary"].between(
        65000,
        90000
    )
]
```

```python
# Sort employees by salary (Lowest -> Highest)

result = df.sort_values(by="salary")

result
```

```python
# Sort employees by salary (Highest -> Lowest)

result = df.sort_values(
    by="salary",
    ascending=False
)

result
```

```python
# Sort by department first, then salary

result = df.sort_values(
    by=["department", "salary"]
)

result
```
